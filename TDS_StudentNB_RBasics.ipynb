Calculating roots of a quadratic in standard form (dear god.)

```r
a <- as.numeric(readline(prompt = "input your a value here (coefficient of x^2)"))
b <- as.numeric(readline(prompt = "input your b value here (coefficient of x)"))
c <- as.numeric(readline(prompt = "input your c value here (constant)"))
root1 <- (-b+sqrt(b^2-4*a*c))/(2*a)
root2 <- (-b-sqrt(b^2-4*a*c))/(2*a)

root1
root2
```

subsetting == allows us to select specific portions of our data
df_name[rows, columns]

viewing specific rows of a dataframe:
df_name[x,] (NOTE: leaving columns blank assumes you want all present)

viewing rows 2 through 10: iris[2:10,]

Viewing rows 2, 6, and 25: iris[c(2,6,25),]

Rows that meet a certain condition: df_name[df_name$column=="specific category",]

specific columns: iris[,c("Species","Sepal.Length")]

Combine to view first 100 rows, but only for species and sepal length: iris[1:100, c("Species","Sepal.Length")]


```r
iris[1,1]
iris[1,]
iris[2:10,]
iris[c(1,3,5),]
iris[iris$Species=="setosa",]
iris[,c("Species","Sepal.Length")]
iris[1:100,c("Species","Sepal.Length")]
```
