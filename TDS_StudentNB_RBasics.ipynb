dplyr is powerful in part due to its simplicity. Here are the main terms you must be familiar with while working in this package:



dplyr Verbs: dplyr verbs are functions that help us change our data. They take in old dataframes as input, and output a new dataframe.

%>% (“pipe”): The pipe symbol is used to pass the output of one function to the next, often “connecting” dplyr verbs.

dplyr Verbs: filter()
filter() = used to keep or discard rows in a dataframe that meet a specified criteria. Remember that conditions can test equality (==) and inequality (<,>,<=,>=). You can combine multiple statements using AND (&) or OR (|)


```r
#Write code to find all diamonds that are at least 1.5 carats but cost less than $4,000.

filteredDiamonds <- diamonds %>%
  filter(carat>=1.5 & price<4000)

filteredDiamonds

#IMPORTANT: Use a new variable name to put the modified stuff into!
```

dplyr Verbs: select()
The select() function is used to keep or discard columns in a dataframe.

```r
#Write code to find all diamonds that are at least 1.5 carats but cost less than $4,000.

selectedDiamonds <- diamonds %>%
  select(carat, price)

selectedDiamonds
```

dplyr Verbs: arrange()
The arrange() function is used to arrange the rows in your dataframe by a single numeric variable.

IMPORTANT: By default, rows are arranged in ascending order. Use the desc() function inside arrange() to switch this.

```r
#Write code to find all diamonds that are at least 1.5 carats but cost less than $4,000.

arrangedDiamonds <- filteredDiamonds %>%
  arrange(price)

arrangedDiamonds
```

The Pipe Symbol: %>%
The pipe symbol can chain together multiple dplyr verbs to faciliate complex data manipulation tasks efficiently. Let’s look under the hood for a second:

Everytime a dplyr verb is applied to a dataframe, a new dataframe is produced.

So, applying a dplyr verb on to another builds upon the changes of the last.



```r
#What takes the casual R programmer 8 lines of code to do

iris1 <- iris %>%
  select(Species, Petal.Length)

iris2 <- iris1 %>%
  filter(Species=="setosa")

iris3 <- iris2 %>%
  arrange(desc(Petal.Length))


#Takes you only 4
iris1 <- iris %>%
  select(Species, Petal.Length) %>%
  filter(Species=="setosa") %>%
  arrange(desc(Petal.Length))
```

dplyr Verbs: summarize()
The summarize() function helps to calculate statistics related to your data. Before diving into its usage, here are some important helper functions:

mean(): Calculates the mean, takes in a vector as input, outputs a single number.
Ex: mean(c(1,2,3,4, NA), na.rm=T)
This code calculates the mean of the specified string of numbers, and removes any NAs before the final calculation.

median(): Calculates the median, takes in a vector as input, outputs a single number.

sd(): Calculates the standard deviation, takes in a vector as input, outputs a single number.

IQR(): Calculates the interquartile range, takes in a vector as input, outputs a single number.

max(): Returns the maximum value in a vector.

min(): Returns the minimum value in a vector.

n(): Counts the number of rows in a group (or whole dataset).

With our helper functions, we are able to summarize our data empirically:

```r
meanBills <- penguins %>%
  select(bill_length_mm,bill_depth_mm)%>%
  summarize(
    mean_length = mean(bill_length_mm,na.rm=T),
    mean_depth=mean(bill_depth_mm, na.rm=T))

meanBills
```

dplyr Verbs: group_by()
The group_by() function lets us manipulate our data by group. This means, that we can do things like summaraize and mutate for specific types of observations.

```r
meanBills <- penguins %>%
  select(bill_length_mm,bill_depth_mm,species)%>%
  group_by(species)%>%
  summarize(
    mean_length = mean(bill_length_mm,na.rm=T),
    mean_depth=mean(bill_depth_mm, na.rm=T))

meanBills
```
