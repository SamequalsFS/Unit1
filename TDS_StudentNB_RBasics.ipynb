Programming language == a set of rules and syntax that dictate how computer programs, or instructions we give to computing devices, are written
Programming languages differ by...
    Syntax == specific structures and cominations of symbols that create meaningful code
    Abstraction == to what degree you are talking directly with the hardware of your computer (memory, CPU registers). The more abstracted, the easier it is for humans to work with. The less abstracted, the more control you have over efficiency.
    Purpose == different fields use languages for specific purposesdue to their particular strengths

R == high-level programming language primarily used by and for data practitioners. It makes it easy for people who work with information to do the 5 primary tasks of data science:
    Data acquisition
    Data wrangling
    Exploration & visualization
    Modeling & inference
    Communicating findings

Calculating roots of a quadratic in standard form (dear god.)

```r
a <- as.numeric(readline(prompt = "input your a value here (coefficient of x^2)"))
b <- as.numeric(readline(prompt = "input your b value here (coefficient of x)"))
c <- as.numeric(readline(prompt = "input your c value here (constant)"))
root1 <- (-b+sqrt(b^2-4*a*c))/(2*a)
root2 <- (-b-sqrt(b^2-4*a*c))/(2*a)

root1
root2
```
